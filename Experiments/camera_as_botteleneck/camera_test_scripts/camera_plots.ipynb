1) Under Default conditions, for each allowed input resolution,  
    (a) Exposure Time analysis

```python
import matplotlib.pyplot as plt
import numpy as np

resolutions = ["640x480", "1296x972", "1920x1080", "2592x1944"]

# Keep µs for min/default/runtime
exp_min = np.array([134, 86, 110, 130])
exp_default = np.array([20000, 20000, 20000, 20000])
exp_runtime = np.array([33167, 33239, 33216, 63835])

# Convert ONLY max to ms
exp_max_ms = np.array([4879289, 3066985, 3066979, 3066985]) / 1000

x = np.arange(len(resolutions))

fig, (ax_max, ax_mid, ax_min) = plt.subplots(
    3, 1, sharex=True, figsize=(6,8),
    gridspec_kw={'height_ratios': [1, 1, 1]}
)

# -------- MAX REGION (ms) --------
ax_max.scatter(x, exp_max_ms, marker='v', label="Max (ms)")
ax_max.set_ylim(2800, 5500)
ax_max.set_ylabel("ExposureTime (ms)")
ax_max.set_title("ExposureTime range (AeEnable = True)")

for i, v in enumerate(exp_max_ms):
    ax_max.text(i, v + 200, f"{v:.1f}", ha='center', va='bottom')

# -------- MID REGION (µs) --------
ax_mid.scatter(x, exp_default, marker='s', label="Default (µs)")
ax_mid.scatter(x, exp_runtime, marker='o', label="Runtime (µs)")
ax_mid.set_ylim(15000, 100000)
ax_mid.set_ylabel("ExposureTime (µs)")

for i, v in enumerate(exp_default):
    ax_mid.text(i, v + 3000, f"{int(v)}", ha='center', va='bottom')

for i, v in enumerate(exp_runtime):
    ax_mid.text(i, v + 3000, f"{int(v)}", ha='center', va='bottom')

# -------- MIN REGION (µs) --------
ax_min.scatter(x, exp_min, marker='^', label="Min (µs)")
ax_min.set_ylim(0, 200)
ax_min.set_ylabel("ExposureTime (µs)")
ax_min.set_xlabel("Input Resolution")

for i, v in enumerate(exp_min):
    ax_min.text(i, v + 10, f"{int(v)}", ha='center', va='bottom')

ax_min.set_xticks(x)
ax_min.set_xticklabels(resolutions, rotation=45)

# Hide connecting spines
ax_max.spines['bottom'].set_visible(False)
ax_mid.spines['top'].set_visible(False)
ax_mid.spines['bottom'].set_visible(False)
ax_min.spines['top'].set_visible(False)

ax_max.tick_params(axis='x', which='both', bottom=False, labelbottom=False)
ax_mid.tick_params(axis='x', which='both', bottom=False, labelbottom=False)

# Add legends
handles1, labels1 = ax_max.get_legend_handles_labels()
handles2, labels2 = ax_mid.get_legend_handles_labels()
handles3, labels3 = ax_min.get_legend_handles_labels()

ax_max.legend(handles1 + handles2 + handles3,
           labels1 + labels2 + labels3,
           loc="upper right")


plt.tight_layout()
plt.show()
```

    (c) FPS analysis

```python
import matplotlib.pyplot as plt
import numpy as np

resolutions = ["640x480", "1296x972", "1920x1080", "2592x1944"]

# FrameDuration in µs
frame_min = np.array([16971, 21581, 30483, 63965])
frame_default = np.array([33333, 33333, 33333, 33333])
frame_runtime = np.array([33302, 33326, 33326, 63965])
frame_max = np.array([4879899, 3067365, 3067365, 3067365])

# Convert to FPS
fps_min = 1e6 / frame_min
fps_default = 1e6 / frame_default
fps_runtime = 1e6 / frame_runtime
fps_max = 1e6 / frame_max

x = np.arange(len(resolutions))

fig, (ax_top, ax_low) = plt.subplots(
    2, 1, sharex=True, figsize=(6,7),
    gridspec_kw={'height_ratios': [2, 1]}
)

# -------- TOP REGION (High + Mid FPS) --------
ax_top.scatter(x, fps_min, marker='^', label="Max (from Min FrameDuration)")
ax_top.scatter(x, fps_default, marker='s', label="Default")
ax_top.scatter(x, fps_runtime, marker='o', label="Runtime")

ax_top.set_ylim(10, 70)
ax_top.set_ylabel("FPS")
ax_top.set_title("Derived FPS range from FrameDuration (AeEnable = True)")

for i, v in enumerate(fps_min):
    ax_top.text(i, v + 1, f"{v:.2f}", ha='center', va='bottom')

for i, v in enumerate(fps_default):
    ax_top.text(i, v - 4, f"{v:.2f}", ha='center', va='bottom')

for i, v in enumerate(fps_runtime):
    ax_top.text(i, v + 1, f"{v:.2f}", ha='center', va='bottom')

# -------- LOW FPS REGION --------
ax_low.scatter(x, fps_max, marker='v', label="Min (from Max FrameDuration)")
ax_low.set_ylim(0, 1)
ax_low.set_ylabel("FPS")
ax_low.set_xlabel("Input Resolution")

for i, v in enumerate(fps_max):
    ax_low.text(i, v + 0.05, f"{v:.2f}", ha='center', va='bottom')

ax_low.set_xticks(x)
ax_low.set_xticklabels(resolutions, rotation=45)

# Remove extra x ticks
ax_top.tick_params(axis='x', bottom=False, labelbottom=False)

# Hide spine between regions
ax_top.spines['bottom'].set_visible(False)
ax_low.spines['top'].set_visible(False)

# Add legends
handles1, labels1 = ax_top.get_legend_handles_labels()
handles2, labels2 = ax_low.get_legend_handles_labels()

ax_top.legend(handles1 + handles2, labels1 + labels2, loc="upper right")

plt.tight_layout()
plt.show()
```
